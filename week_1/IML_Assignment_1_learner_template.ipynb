### a) Discuss the implications that uncertainty in $X$ would have on parameter estimates based on linear regression models. Should we care about error in $X$ when fitting OLS regressions?



Provide your answer in this Markdown cell

#### Implications on parameter estimates because of uncertainty in X:

1. Uncertainty in X can add bias to the parameters-intercept and predictor variables. Moreover, it can add potential variance terms to the parameters.
2. This leads to inaccurate parameter estimates that effect the predictions accuracy. This increases the width of confidence intervals for the parameter estimates.

#### Significance of error in X:

1. We should include the error in X when fitting OLS regressions because uncertainty in X causes inaccurate predictions by adding bias to the model.
2. We cannot capture the exact relationship between the predictors and the response if we ignore the error in X.

### b) Provide a brief discussion on the concept of regression dilution.




Provide your answer in this Markdown cell

1. Regression dilution refers to the concept where the linear regression parameter estimates are biased towards zero because of an error in the measurement of predictors or presence of uncertainty in the predictors
2. Regression dilution can occur because of the presence of variability in the predictors. This can happen because of imperfect data collection, instrument error and irreducible variability in the data.
3. Because of this measurement error in the predictors, the true relation between the predictors and response cannot be learnt by the model resulting in clipping or attenuation of parameter estimates i.e., biasing them towards zero.
4. Regression dilution can be addressed by careful measurements of variables and modern regression techniques that are robust to these kind of errors.

### d) Distinguish between least squares and maximum likelihood in the context of linear regression models



Provide your answer in this Markdown cell

1. The least squares method minimizes the residual sum of squares whereas maximum likelihood method maximizes the likelihood function.
2. The least squares method focuses on finding the line that fits the data well by minimizing the RSS whereas maximum likelihood estimation finds the parameter estimates using a probabilistic framework by maximizing the likelihood function.
3. The least squares method is unbiased whereas maximum likelihood method is unbiased in large sample sizes only. The maximum likelihood method is biased in small sample sizes.

```python
#IMPORT LIBRARIES
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

```python
#READ AUTO.CSV DATASET
org = pd.read_csv('./Auto.csv')
df = org.copy()
df.head()
```

```python
df.info()
```

```python
hp_list = list(df['horsepower'])
for idx,item in enumerate(hp_list):
    if item=='?':
        print(f'Index: {idx} and corresponding value: {item}')
```

```python
#PERFORM PRE PROCESSING OF DATA IF REQUIRED ( OBSERVE THE DATASET AND ITS VALUES CAREFULLY)

# horsepower column is an object datatype but it should be an int datatype. So, converting horsepower to int datatype
# Before converting it into int datatype some of the entries has '?' value. Let's impute it using the median value.
# We will perform the following the steps:

# Replace '?' with NaN
df['horsepower'] = df['horsepower'].replace('?', pd.NA)

# Convert 'horsepower' column to numeric
df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')

median = df['horsepower'].median()
df['horsepower'] = df['horsepower'].fillna(median) # fill with median value

#df['horsepower'] = df['horsepower'].astype(int)

df.info() # now horsepower column's datatype is float.
```

```python
Y = df['mpg']
X = sm.add_constant(df['horsepower'])

simple_line_reg = sm.OLS(Y,X).fit()

print(simple_line_reg.summary())
```

### e) Fit a linear regression model with interaction effects. explore a few different models (~3–4 models). Do any interactions appear to be statistically significant in any of the models you examined?

Use formula.ols to check the interaction between two predictors as used in the below reference link

https://www.statsmodels.org/devel/example_formulas.html

```python
# BEGIN SOLUTION
import statsmodels.formula.api as smf

new_df = df[['mpg','displacement','weight','year','origin']] # these columns or predictors are choosen because they have significant relationship to the response mpg

model_expresion1 = 'mpg ~ displacement*weight+1'
interaction_model1 = smf.ols(formula=model_expresion1,data=new_df).fit()
print('Interaction between displacement & weight\n\n',interaction_model1.summary(),'\n\n\n\n')

model_expresion2 = 'mpg ~ year*origin+1'
interaction_model2 = smf.ols(formula=model_expresion2,data=new_df).fit()
print('Interaction between year & origin\n\n',interaction_model2.summary(),'\n\n\n\n')

model_expresion3 = 'mpg ~ displacement*origin+1'
interaction_model3 = smf.ols(formula=model_expresion3,data=new_df).fit()
print('Interaction between displacement & origin\n\n',interaction_model3.summary(),'\n\n\n\n')

model_expresion4 = 'mpg ~ year*weight+1'
interaction_model4 = smf.ols(formula=model_expresion4,data=new_df).fit()
print('Interaction between year & weight\n\n',interaction_model4.summary())
# END SOLUTION
```

1. Out of the 4 models with interaction effects examined in the above cell, 3 of the interactions have a significant relationship to the response mpg.
2. The interactions displacement*weight, displacement*origin and year*weight have a significant relationship with the response mpg.
3. The interaction year*origin does not have any significant relationship with the response mpg.

## Question 8

It is claimed that in the case of simple linear regression of $Y$ onto $X$, the $R^2$ statistic is equal to the square of the correlation between $X$ and $Y$. Generate some synthetic data and show that this is the case.

link to  correlation function:
https://numpy.org/doc/stable/reference/generated/numpy.corrcoef.html

```python
# BEGIN SOLUTION
np.random.seed(8)
gen_x = np.random.rand(100)
gen_y = -3 + 3*gen_x + np.random.normal(0,0.3,100)

correlation = np.corrcoef(gen_x, gen_y)[0,1]

gen_X = sm.add_constant(gen_x)
gen_model = sm.OLS(gen_y,gen_X).fit()

formula_r_square = correlation**2
gen_model_r_square = gen_model.rsquared

print(f"R-squared from the formula = {round(formula_r_square,4)} and R-squared from the model = {round(gen_model_r_square,4)}")

if round(formula_r_square,4)==round(gen_model_r_square,4):
    print("Hence, R-squared is equal to the square of the correlation between X and Y for a simple linear regression of Y onto X")
# END SOLUTION
```
