Q2. Suppose we produce ten bootstrapped samples from a data set containing red and green classes. We then apply a classification tree to each bootstrapped sample and, for a specific value of $X$, produce $10$ estimates of $P(Class is Red|X)$:

$0.1$, $0.15$, $0.2$, $0.2$, $0.55$, $0.6$, $0.6$, $0.65$, $0.7$, and $0.75$.

There are two common ways to combine these results together into a single class prediction. One is the majority vote approach , the second approach is to classify based on the average probability. In this example, what is the final classification under each of these two approaches?

ANSWER in the markdown cell

1. We have ten bootstrapped samples with us and a classification tree is fit to each bootstrapped sample. For a given value of X, the 10 estimates of $P(Class is Red|X)$ are $0.1$, $0.15$, $0.2$, $0.2$, $0.55$, $0.6$, $0.6$, $0.65$, $0.7$, and $0.75$.

2. Using the majority vote approach, 6 out of 10 estimates($0.55$, $0.6$, $0.6$, $0.65$, $0.7$, and $0.75$) have $P(Class is Red|X)>0.5$ indicating that the observation X belongs to class 'Red' whereas 4 out of 10 estimates($0.1$, $0.15$, $0.2$, $0.2$) have $P(Class is Red|X)<0.5$ indicating that it belongs to class 'Green'. In majority vote approach, we choose the class with the highest number of votes supporting the class. Therefore, according to majority vote approach, the observation belongs to the class 'Red'.

3. In average probability approach, the observation is classified based on the average of estimates or probabilities. The average probability of $0.1$, $0.15$, $0.2$, $0.2$, $0.55$, $0.6$, $0.6$, $0.65$, $0.7$, and $0.75$ is 0.45(<0.5) which is less than 0.5. This indicates that the observation does not belong to class 'Red'. In otherwords, it belongs to the class 'Green'.

4. Hence, according to majority vote approach the observation X belongs to the class 'Red' whereas according to average probability approach it belongs to the class 'Green'.

```python
# import libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_squared_error, accuracy_score
from ISLP import load_data
```

#### a) Read the dataset using function read_csv( ) from pandas library and preprocess the data if required ( check for any missing values etc )



To answer the following questions Auto.csv dataset from week-1 is used. The dataset can be loaded either using pd.read_csv() or using load_data() from ISLP package. For now pd.read_csv() method is used.

```python
org  = pd.read_csv('./Auto.csv')
org.info()
org.head()
```

```python
df = org.copy()
print(df['horsepower'].unique()) # There is a '?' present in the horsepower column.

non_num_indices = []
for idx, hp in enumerate(list(df['horsepower'])):

    if hp=='?':
        non_num_indices.append(idx)
        print(f'Index {idx} has a value = {hp}')
non_num_indices
        
```

```python
# Preprocessing - remove rows from the dataset with '?' value in the horsepower column

df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')

df = df.dropna(subset=['horsepower'])

df.info()
df.head()
```

```python
num_df = df.select_dtypes(include=['float64','int64']) # select numeric columns only

corr_matrix = num_df.corr()


sns.heatmap(corr_matrix,annot=True)
plt.title('Heatmap')
plt.show()
corr_matrix
```

```python
sns.pairplot(df)
plt.show()
```

**We will predict mpg using other variables**

#### b) Split the data into train data  and test data ( proportion of train and test data of your choice )

Reference :

https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.train_test_split.html



```python
y = num_df['mpg']
X = num_df.drop(columns=['mpg'])

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, random_state=12345)
```

```python
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

#### c) Train the data on the below models

i) Linear regression model
ii) Gradient boosting model
iii) Random forest model

check the train and test r2 score and mention your findings on which model performed better in terms of underfitting / overfitting etc

References:

https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html

https://scikit-learn.org/stable/auto_examples/ensemble/plot_gradient_boosting_regression.html#gradient-boosting-regression

https://scikit-learn.org/stable/modules/generated/sklearn.ensemble.RandomForestRegressor.html

https://scikit-learn.org/stable/modules/generated/sklearn.metrics.r2_score.html

**(i) Fitting Linear Regression Model**

```python
lr_model = LinearRegression().fit(X=X_train, y=y_train)

lr_train_r2 = lr_model.score(X_train,y_train)
lr_y_test_pred = lr_model.predict(X_test)
lr_test_r2 = r2_score(y_test,lr_y_test_pred)

lr_train_mse = mean_squared_error(y_train,lr_model.predict(X_train))
lr_test_mse = mean_squared_error(y_test,lr_y_test_pred)

print('R2 score for Linear Regression model on train dataset is',round(lr_train_r2,4),'\n')

print('R2 score for Linear Regression model on test dataset is',round(lr_test_r2,4) ,'\n')

print(f'MSE for Linear Regression model on train dataset = {round(lr_train_mse,4)}\n')
print(f'MSE for Linear Regression model on test dataset = {round(lr_test_mse,4)}\n')
```

**(ii) Fitting Gradient Boosting Model**

```python
gb_model = GradientBoostingRegressor().fit(X=X_train,y=y_train)

gb_train_r2 = gb_model.score(X_train,y_train)
gb_y_test_pred = gb_model.predict(X_test)
gb_test_r2 = r2_score(y_test,gb_y_test_pred)

gb_train_mse = mean_squared_error(y_train,gb_model.predict(X_train))
gb_test_mse = mean_squared_error(y_test,gb_y_test_pred)

print('R2 score for Gradient Boosting Regression model on train dataset is',round(gb_train_r2,4),'\n')

print('R2 score for Gradient Boosting Regression model on test dataset is', round(gb_test_r2,4),'\n')

print(f'MSE for Gradient Boosting Regression model on train dataset = {round(gb_train_mse,4)}\n')
print(f'MSE for Gradient Boosting Regression model on test dataset = {round(gb_test_mse,4)}\n')
```

```python
gb_feature_imp = gb_model.feature_importances_
sorted_indices = gb_feature_imp.argsort()[::-1]
sorted_feature_names = [X.columns[i] for i in sorted_indices]
sorted_feat_imp = gb_feature_imp[sorted_indices]

plt.figure()
plt.barh(sorted_feature_names,sorted_feat_imp)
plt.title('Feature Importance for Gradient Boosting Regression Model')
plt.gca().invert_yaxis()
plt.show()
```

**(iii) Random Forest Regressor**

```python
rf_model = RandomForestRegressor().fit(X=X_train,y=y_train)

rf_train_r2 = rf_model.score(X_train,y_train)
rf_y_test_pred = rf_model.predict(X_test)
rf_test_r2 = r2_score(y_test,rf_y_test_pred)

rf_train_mse = mean_squared_error(y_train,rf_model.predict(X_train))
rf_test_mse = mean_squared_error(y_test,rf_y_test_pred)

print('R2 score for Random Forest Regression model on train dataset is',round(rf_train_r2,4),'\n')

print('R2 score for Random Forest Regression model on test dataset is', round(rf_test_r2,4),'\n')

print(f'MSE for Random Forest Regression model on train dataset = {round(rf_train_mse,4)}\n')
print(f'MSE for Random Forest Regression model on test dataset = {round(rf_test_mse,4)}\n')
```

```python
rf_feature_imp = rf_model.feature_importances_
sorted_indices = rf_feature_imp.argsort()[::-1]
sorted_feature_names = [X.columns[i] for i in sorted_indices]
sorted_feat_imp = rf_feature_imp[sorted_indices]

plt.figure()
plt.barh(sorted_feature_names,sorted_feat_imp)
plt.title('Feature Importance for Random Forest Regression Model')
plt.gca().invert_yaxis()
plt.show()
```

1. The $R^{2}$ values for the Linear regression model on training and test dataset are 0.8276 and 0.8005 respectively.

2. The $R^{2}$ values for the Gradient boosting regression model on training and test dataset are 0.9767 and 0.8661 respectively.

3. The $R^{2}$ values for the Random Forest regression model on training and test dataset are 0.9809 and 0.8735 respectively.

4. From the above observed $R^{2}$ values it is clear that Random forest regression model is more generalized as it has higher $R^{2}$ value for the test dataset among all the models.

5. The MSE values on the test dataset for Linear regression, gradient boosting regression and random forest regression models are 12.6989, 8.5255 and 8.0531 respectively. These values indicate that the random forest regression model is better than the other two models because it has the smallest test set MSE.

6. Linear regression model seems to underfit the data as it has the lowest $R^{2}$ value on both training and test datasets among all the models.

7. Random forest regression and Gradient boosting regression models appear to overfit the data as they have a large difference between the training and test set $R^{2}$ values. Moreover, random forest model has the largest difference between the training and test set $R^{2}$ values.

8. Even though Random forest and Gradient boosting models overfit the data, they are still better than the linear regression model as they perform better on the test dataset.

9. Finally, Random forest is the best model as it has lower test set MSE and higher test set $R^{2}$ values among all the three models. Additionally, it learns the underlying patterns in the data better than the remaning two.
