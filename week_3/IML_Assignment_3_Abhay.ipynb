## Question 1

Suppose that a curve $\hat{g}$ is computed to smoothly fit a set of `n` points using the following formula:

  $$
  \hat{g} = \arg\min_{g}(\sum_{i=1}^n (y_i - g(x_i))^2 +  \lambda \int [g^{(m)} (x)]^2 dx )
  $$
where $g^{(m)}$  represents the $m^{th}$ derivative of $g$ (and $g^{(0)} = g$. Provide a description (or sketch) of $\hat{g}$ in each of the following scenarios. where $g^{(m)}$
 represents the m
th derivative of g



  (a) $\lambda$ = $\infty$, $m = 0$.
  

  


**ANSWER**

1. When $\lambda=\infty$ and m=0, the regularization term  $\lambda \int [g^{(m)} (x)]^2 dx )$ dominates and the first term can be neglected.

2. Then the given equation reduces to $  \arg\min_{g}\lambda \int [g(x)]^2 dx  $ . To minimize this term with $\lambda =\infty$, g(x) must be equal to zero.
  
3. Hence, $\hat g = 0$ when $\lambda=\infty$ and m=0


  (b) $\lambda$ = $\infty$, $m = 1$.
  



**ANSWER** 

1. When $\lambda=\infty$ and m=1, the regularization term  $\lambda \int [g^{(m)} (x)]^2 dx )$ dominates and the first term can be neglected.
   
2. Then the given equation reduces to $  \arg\min_{g}\lambda \int [g^{1}(x)]^2 dx  $ . To minimize this term with $\lambda =\infty$, $g^{1}(x)$ must be equal to zero.

3. Therefore, $g^{1}(x)=g'(x)=0$. This implies that g(x) is a constant function as derivative of a constant function is zero

4. Hence, g(x) is a constant c. $\hat g = c$  when $\lambda=\infty$ and m=1.


  
  (c) $\lambda$ = $\infty$, $m = 2$.
  

  


**ANSWER**

1. When $\lambda=\infty$ and m=2, the regularization term  $\lambda \int [g^{(m)} (x)]^2 dx )$ dominates and the first term can be neglected.
   
2. Then the given equation reduces to $  \arg\min_{g}\lambda \int [g^{2}(x)]^2 dx  $ . To minimize this term with $\lambda =\infty$, $g^{2}(x)$ must be equal to zero.

3. Therefore, $g^{2}(x)=g"(x)=0$. This implies that g'(x) is a constant function as derivative of a constant function is zero. g'(x) = c implies that g(x) is a linear function of x.

4. Hence, g'(x) = c implies that $g(x) = cx + d$ when $\lambda=\infty$ and m=2.

  (d) $\lambda$ = $\infty$, $m = 3$.




**ANSWER**

1. When $\lambda=\infty$ and m=3, the regularization term  $\lambda \int [g^{(m)} (x)]^2 dx )$ dominates and the first term can be neglected.
   
2. Then the given equation reduces to $  \arg\min_{g}\lambda \int [g^{3}(x)]^2 dx  $ . To minimize this term with $\lambda =\infty$, $g^{3}(x)$ must be equal to zero.

3. Therefore, $g^{3}(x)=0$. This implies that $g^{2}(x)=g"(x)$ is a constant function as derivative of a constant function is zero. g"(x) = c implies that g'(x) is a linear function of x and g(x) is a quadratic function of x.

4. g"(x) = c implies that $g'(x) = cx + d$. $g'(x) = cx+d$ implies that $g(x) = cx^{2} + dx + e $ when $\lambda=\infty$ and m=3.

  
  (e) $\lambda$ = 0, $m = 3$.

**ANSWER**

1. When  $\lambda$ = 0, $m = 3$, the regularization term becomes zero or vanishes.

2. Then the given equation reduces to $\arg\min_{g}(\sum_{i=1}^n (y_i - g(x_i))^2)$. The problem reduces to finding $\hat g$ that minimizes the residual sum of squares $\sum_{i=1}^n (y_i - g(x_i))^2$.

3. This implies that $\hat g $ should interpolate each and every $y_{i}$. It is equivalent to finding the ordinary least squares(OLS) fit for linear regression.

## Question 2

For (1) step functions, (2) local regression, and (3) GAMs: provide one situation when using that particular model is advantageous. For instance, “GAMs is likely the preferred option when the question I’m interested in addressing involves…”. 


**ANSWER**

1. Step functions can be used in scenarios where we do not want fit a single non-linear function on the entire range of X. It is mainly advantageous when categorical predictors are involved in predicting the response. Step-functions are preferred when we want to effectively capture a sudden change in the response based on the impact of different levels of predictors. For instance, step-functions can be used to predict the response wage based on the different levels or categories of age. Categorical predictors can also be created by binning the continuous predictor values and step-function approach can be carried out to predict the response.

2. Local regression can be used in scenarios where there is a need to fit a flexible non-linear function. It involves computing the fit a target point using only the neighbouring observations. Here, weighted least squares linear regression is preformed at each target point by selecting the neighbouring points. It is preferred in scenarios where we want to predict the response at a point based on the impact of the points nearby. For instance, it can be used to predict the interest rates on a stock based on the stock prices, exchange rates, and other economic indicators. We get good estimates of the interest rates at a given time using local regression because the values are impacted mainly by the predictor values  in recent times.

3. GAMs can be used in scenarios where the response is estimated using non-linear functions of each of the variables or predictors while maintaining additivity. In GAMs a separate non-linear function($f_{j}$) is calculated for each predictor($X_{j}$) and then all are added together. It also maintains interpretability. GAMs are preferred in scenarios where we want to capture non-linear relationships between the predictors and the response, and to examine the effect of each $X_{j}$ on Y individually while holding all of the other variables fixed. For example, they can be used to predict the health of an individual based on different variables such as temperature, dust level, air pollution and carbon emissions.

## Question 3 

Define the following types of smooth functions from PYGAM library (https://pygam.readthedocs.io/en/latest/notebooks/tour_of_pygam.html ) and list at least one relevant potential use: 

a) l() linear terms

b) s() spline terms

c) f() factor terms

d) te() Tensor product interaction terms


**ANSWER**

a) l() linear terms: This function is used to model a simple linear relation between the predictor variable and the response. By default it has l2 ridge penalty. The main arguments of the function are feature index, tuning or smoothing parameter and type of smoothing penalty(L2, derivative etc.,). For example one potential use case is to predict the house price based only on the number of rooms.

b) s() spline terms: This function is used when the relationship between the predictor variable and the response is non-linear. They can be used to fit various non-linear functions. By default it has a penalty on its 2nd derivative and uses 20 splines. The main arguments of the function are feature index, tuning or smoothing parameter, type of smoothing penalty(L2, derivative etc.,), order of the spline and type of basis functions. For example one potential use case is to model the non-linear relationship between the risk of diabetes and hours of exercise. Another use case is to predict the risk of lung allergy based on the time exposed to pollutants.

c) f() factor terms: This function is used to model the relationship between categorical predictors and the response. By default it has l2 ridge penalty. The main arguments of the function are feature index, tuning or smoothing parameter, type of smoothing penalty and type of encoding for feature categories(one-hot). For example one potential use case is to predict the income based on the gender. The factor terms can effectively capture the underlying relation between the income and the categories of gender.

d) te() Tensor product interaction terms: This function is used to model the underlying interactions between two or more predictors and the response. The main arguments of the function are feature index, tuning or smoothing parameter, type of smoothing penalty, number of splines and order of the spline. For example one potential use case is to model the effect of interaction between temperature and humidity and crop yield.








## Question 4

Briefly explain what spatial and temporal autocorrelation are? List one way in which GAMs could help to account for this (e.g. indicate the type of smooth function you would use or potential types of correlation structures).



**ANSWER**

1. Spatial autocorrelation is the correlation between points or observations at different locations in space. Points closer to each other are more similar or correlated than the points that are farther apart. GAMs account for spatial autcorrelation by adding latitude and longitude as smoothing terms. Thin-plate spline function in GAM is used as a smooth function to capture the variability in space or to account for spatial autocorrelation.

2. Temporal autocorrelation is the correlation between observations at different points in time. Observations that are closer in time are more similar or correlated than those that are farther apart. Cyclic cubic regression splines are used as smoothing terms in GAMs to account for the temporal autocorrelation. Apart from this, moving average correlation structure is also used to account for temporal autocorrelation.

Consider two curves $\hat{g}1$ and $\hat{g}2$, defined by

  $$
  \hat{g}1 = \arg\min_{g}(\sum_{i=1}^n (y_i - g(x_i))^2 +  \lambda \int [g^{(3)} (x)]^2 dx )
  $$
  
  $$
  \hat{g}2 = \arg\min_{g}(\sum_{i=1}^n (y_i - g(x_i))^2 +  \lambda \int [g^{(4)} (x)]^2 dx )
  $$
  
 
where $g^{(m)}$ represents the $m^{th}$ derivative of $g$. Now, answer the following questions (and include a brief explanation):

  a) As $\lambda \rightarrow \infty$, will $\hat{g}1$ or $\hat{g}2$ have smaller training RSS?
  

**ANSWER**

1. When $\lambda=\infty$, only the second terms of $\hat g1$ and $\hat g2$ dominate and the first terms can be ignored. The equations reduce to $\arg\min_{g} \lambda \int [g^{(3)} (x)]^2 dx$ and $\arg\min_{g} \lambda \int [g^{(4)} (x)]^2 dx$.

2. In case of $\hat g1$, the equation reduces to $\arg\min_{g} \lambda \int [g^{(3)} (x)]^2 dx$. To minimize this term with $\lambda=\infty$, $g^{3}(x)$ must be zero. Therefore, $g^{3}(x)=0$. This implies that $g^{2}(x)=g"(x)$ is a constant function as derivative of a constant function is zero. g"(x) = c implies that g'(x) is a linear function of x and g(x) is a quadratic function of x. g"(x) = c implies that $g'(x) = cx + d$. $g'(x) = cx+d$ implies that $g(x) = cx^{2} + dx + e$. Hence, $\hat g1$ is a quadratic polynomial.

3. In case of $\hat g2$, the equation reduces to $\arg\min_{g} \lambda \int [g^{(4)} (x)]^2 dx$. To minimize this term with $\lambda=\infty$, $g^{4}(x)$ must be zero. Therefore, $g^{4}(x)=0$. This implies that $g^{3}(x)$ is a constant function as derivative of a constant function is zero. $g^{3}(x)=c$ implies that g"(x) is a linear function of x and g'(x) is a quadratic function of x. $g"(x) = cx + d$ implies that $g'(x) = cx^{2} + dx + e$. $g'(x) = cx^{2} + dx + e$ implies that $g(x) = cx^{3} + dx^{2} + ex + f$. Hence, $\hat g2$ is a cubic polynomial.

4. $\hat g2$ is more flexible than $\hat g1$ as $\hat g2$ is a cubic polynomial and $\hat g1$ is a quadratic polynomial. $\hat g2$ being a cubic polynomial can fit the training data better than $\hat g1$(quadratic) and can capture the underlying patterns more effectively than $\hat g1$. Hence, $\hat g2$ has smaller training RSS than $\hat g1$.

  b) As $\lambda \rightarrow \infty$, will $\hat{g}1$ or $\hat{g}2$ have smaller test RSS?

**ANSWER**

1. We cannot determine whether $\hat g1$ or $\hat g2$ has smaller test RSS because we do not know the information regarding the degree of non-linearity in the true relationship between the predictor and the response.

2. If the true relation between predictor and response is closer to a quadratic relationship then $\hat g1$ has lower test RSS.

3. If the true relation between predictor and response is closer to a cubic relationship then $\hat g2$ has lower test RSS.

  
  c) For $\lambda= 0$, will $\hat{g}1$ or $\hat{g}2$ have smaller training RSS?

**ANSWER**

1. When $\lambda=0$, both $\hat g1$ and $\hat g2$ reduce to $\arg\min_{g}(\sum_{i=1}^n (y_i - g(x_i))^2)$. The problem reduces to minimizing the residual sum of squares $\sum_{i=1}^n (y_i - g(x_i))^2$.

2. Both $\hat g1$ and $\hat g2$ are same and they will interpolate each and every $y_{i}$. Hence, both will have the same training RSS.

```python
X = df_normalized[['Private','Room_Board','PhD','perc_alumni','Expend','Grad_Rate']]
y = df_normalized['Outstate']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=12345)

```

```python
gam = LinearGAM(f(0) + s(1) + s(2) + s(3) + s(4) + s(5))

# Fit the GAM model
gam.fit(X_train, y_train)
gam = gam.gridsearch(X_train, y_train,lam= np.logspace(-3, 3, 30))
# Plot the results
plt.figure(figsize=(10, 6))

# Plot each predictor against the response variable
for i, term in enumerate(gam.terms):
    if term.isintercept:
        continue
    plt.subplot(2, 3, i+1)
    XX = gam.generate_X_grid(term=i)
    plt.plot(XX[:, i], gam.partial_dependence(term=i, X=XX))
    plt.scatter(X_train.iloc[:, i], y_train, facecolor='red', edgecolors='none', alpha=0.7)
    #plt.title(X_train.columns[i])  
    plt.xlabel(X_train.columns[i])
    plt.ylabel('Outstate')

plt.tight_layout()
plt.show()

print(gam.summary())
```
