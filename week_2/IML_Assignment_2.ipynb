##  Q1. We will derive the probability that a given observation is part of a bootstrap sample. Suppose that we obtain a bootstrap sample from a set of `n` observations. **Please note that samples are obtained with replacement**.

### a) What is the probability that the first bootstrap observation is not the $jth$ observation from the original sample? Justify your answer.

Provide your answer in this Markdown cell

1. We have a total of n observations and we need to pick the first bootstrap observation that is not the $j^{th}$ observation from the original sample.
2. So, the number of observations in the original sample excluding the $j^{th}$ observation is n-1.
3. Probability that the first bootstrap observation is not the $j^{th}$ observation from the original sample is the number of observations in the original sample excluding the $j^{th}$ observation divided by the total number of observations in the original sample.
4. Therefore, the probability is $${n-1}\over {n}$$

###  c) Argue that the probability that the $jth$ observation is not in the bootstrap sample is $(1 − 1/n)^n$.

Provide your answer in this Markdown cell

1. Probability that the $j^{th}$ observation is not in the bootstrap sample is the product of probabilties that the each bootstrap observation is not the $j^{th}$ observation from the original sample.
2. Probability that any $k^{th}$ observation in the bootstrap sample is not the $j^{th}$ observation is the number of observations in the original sample excluding the $j^{th}$ observation divided by the total number of observations in the original sample. This probability is $${n-1}\over{n}$$
3. Now the above probability is multiplied n-times as these are independent events and size of bootstrap sample is same as the size of original sample(n).
4. The probability is $${\prod_{i=1}^{n}} {n-1\over n} =  (1 − 1/n)^n$$

## Q2)
The following questions are relative to k-fold cross-validation.

### a) Explain how k-fold cross-validation is implemented.


Provide your answer in this Markdown cell

1. In k-fold cross-validation, the training dataset is split into k-partitions that do not overlap with each other.
2. In each iteration, 1 partition is used as a validation set and the remaining k-1 partitions are used to fit the model.
3. In each iteration MSE is calculated on that 1 partition or fold used as a validation set. Let $MSE_{i}$ represent the mean squared error in the $i^{th}$ iteration where $i^{th}$ partition is used as a validation set.
4. The average of $MSE_{i}$ over all the k-folds or k-partitions gives the estimate of the test error. It is given by $${1 \over k}\sum_{i=1}^{k}MSE_{i}$$

### b) How would you choose the analyzed k? Does that matter?

Provide your answer in this Markdown cell

1. The value of k depends on factors such as size of the dataset and bias-variance trade-off.
2. If the size of the dataset is small then a large k-value results in an accurate estimate of the test error whereas if the size of the dataset is large then a small k-value results in a good estimate of the test error.
3. On the other hand, a small k-value results in a model with high bias and low variance whereas a large k-value results in a model with low bias and high variance.
4. As k increases the variance increases because of the increase in the overlap of training sets which results in nearly identical observations. On the other hand, a small k value results in a small overlap in training sets.
5. We need to choose the k value keeping in view of the above implications and in general we choose a k value that results in neither high bias nor high variance.
6. Practically, k=5 or k=10 is considered as it results in test error estimate that has neither high bias nor high variance.

### c) What are the advantages and disadvantages of k-fold cross-validation relative to: *The validation set* approach?


Provide your answer in this Markdown cell

1. The validation set approach tends to overestimate the test error as it uses only 50% of the observations to train the model and the remaining 50% are used as a validation set. On the contrary the k-fold cross validation approach uses $(k-1)*n/k$ observations to train the model and $n/k$ observations as validation set(k=5 or 10 in general). In k-fold cross validation, more observations are used to train the model compared to the validation set approach. Hence, k-fold cross validation approach results in a better estimate of test set error compared to the validation set approach.
2. In validation set approach, the model is trained on fewer observations which results in high variability of the test set error estimate. But in k-fold cross-validation, the model is trained on more number of observations which results in decrease in the variability of the test set error estimate.
3. On the other hand, the disadvantage of k-fold cross validation is that it is computationally more expensive than the validation set approach. In validation set approach the model is fit only once whereas in k-fold cross validation the model is fit k-times which increases the number of computations.


### d)  What are the advantages and disadvantages of k-fold cross-validation relative to: *LOOCV*?

Provide your answer in this Markdown cell

1. LOOCV is a special case of k-fold cross-validation in which k is equal to the size of the dataset(k=n). In each iteration, n-1 observations are used to fit the model and the remaining one observation is used for prediction. Here the test set error estimate is given by $${1 \over n}\sum_{i=1}^{n}MSE_{i}$$
2. Each training set in LOOCV approach has n-1 observations, which is nearly all the observations of the dataset. So, it gives nearly unbiased estimates of the test set error. But in k-fold cross-validation(k<n), only $(k-1)*n/k$ observations(fewer than in the LOOCV approach) are used to fit the model. This results in a relatively more biased estimate of test set error compared to the LOOCV approach. Hence, the LOOCV has far less bias compared to k-fold cross validation approach(k<n).
3. The disadvantage is that LOOCV has more variance than k-fold cross validation(k<n) because the model is trained on nearly same data each time and the average prediction error is calculated. Moreover, these prediction errors are highly correlated because the fitted models do not differ in the training data significantly. In k-fold cross validation, each model is trained on a dataset that is relatively more different than in LOOCV approach. Hence, the prediction errors are not as correlated as in LOOCV approach resulting in less variance. Thus, LOOCV has more variance than k-fold cross-validation approach(k<n).
4. The LOOCV is more computationally expensive(except in linear or polynomial models fit by least squares method ) than k-fold cross-validation since in LOOCV the model is fit n-times whereas in k-fold cross-validation the model is fit k-times where k<n.

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import KFold
import matplotlib.pyplot as plt

# Define constants
N = 100
ORDER = 2
N_FOLDS = 20
POLY_DEGREE = np.arange(1, 11)
N_REPLICATES = 10

# Set random seed for reproducibility
np.random.seed(123)

# Generate simulated data
x = np.random.uniform(low=-4, high=4, size=N)
y = np.random.normal(loc=x**ORDER, scale=0.25, size=N)
data = pd.DataFrame({'x': x, 'y': y})

# Run multiple replicates and average the MSE estimates
mse_estimates = np.zeros((N_REPLICATES, len(POLY_DEGREE)))
for r in range(N_REPLICATES):
    kf = KFold(n_splits=N_FOLDS, shuffle=True)
    indices = [(train_index, test_index) for train_index, test_index in kf.split(data)]
    for q_idx, q in enumerate(POLY_DEGREE):
        y_hat = np.zeros(N)
        for train_index, test_index in indices:
            # Fit on K-1 folds
            x_train, y_train = data.iloc[train_index]['x'], data.iloc[train_index]['y']
            poly = PolynomialFeatures(degree=q)
            x_train_poly = poly.fit_transform(x_train.reshape(-1, 1))
            lin_reg = LinearRegression()
            lin_reg.fit(x_train_poly, y_train)

            # Predict on the kth fold
            x_test = data.iloc[test_index]['x']
            x_test_poly = poly.fit_transform(x_test.reshape(-1, 1))
            y_hat[test_index] = lin_reg.predict(x_test_poly)

        # Get the MSE estimate
        mse_estimate = np.mean((y_hat - data['y'])**2)
        mse_estimates[r, q_idx] = mse_estimate

# Average the MSE estimates across replicates
mse_mean = np.mean(mse_estimates, axis=0)
mse_std = np.std(mse_estimates, axis=0)
```

7. After finding, mse_estimate for each replicate and each polynomial degree 'q' mse mean and mse standard deviation are calculated for each polynomial degree.

```python
# BEGIN SOLUTION

import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import KFold
import matplotlib.pyplot as plt

# Define constants
N = 100
ORDER = 2
N_FOLDS = 20
POLY_DEGREE = np.arange(1, 11)

# Set random seed for reproducibility
np.random.seed(123)

# Generate simulated data
x = np.random.uniform(low=-4, high=4, size=N)
y = np.random.normal(loc=x**ORDER, scale=0.25, size=N)
data = pd.DataFrame({'x': x, 'y': y})

# Initialize lists to store MSE mean and standard deviation
mse_mean = []
mse_std = []

# Initialize lists to store coefficients of selected model
selected_model_coefficients = []

# Initialize variable to store selected model degree
selected_model_degree = None

# Perform k-fold cross-validation for each polynomial degree
for degree in POLY_DEGREE:
    kf = KFold(n_splits=N_FOLDS, shuffle=True, random_state=123)
    mse_deg = []
    model_coefficients = []

    for train_index, test_index in kf.split(data):
        x_train, y_train = data.iloc[train_index]['x'], data.iloc[train_index]['y']
        x_test, y_test = data.iloc[test_index]['x'], data.iloc[test_index]['y']

        poly_features = PolynomialFeatures(degree=degree)
        x_train_poly = poly_features.fit_transform(x_train.values.reshape(-1, 1))
        x_test_poly = poly_features.transform(x_test.values.reshape(-1, 1))

        model = LinearRegression()
        model.fit(x_train_poly, y_train)
        y_pred = model.predict(x_test_poly)

        mse_fold = np.mean((y_pred - y_test)**2)
        mse_deg.append(mse_fold)

        if degree == ORDER:
            model_coefficients.append(model.coef_)

    mse_mean.append(np.mean(mse_deg))
    mse_std.append(np.std(mse_deg))

    if degree == ORDER:
        selected_model_degree = degree
        selected_model_coefficients = np.mean(model_coefficients, axis=0)

# Plot MSE vs Polynomial Order
plt.errorbar(POLY_DEGREE, mse_mean, yerr=mse_std, fmt='o-', color='b')
plt.title('MSE vs Polynomial Order')
plt.xlabel('Polynomial Order')
plt.ylabel('Mean Squared Error')
plt.grid(True)
plt.show()

# Plot x vs y and the selected model
plt.scatter(data['x'], data['y'], label='Data')
x_values = np.linspace(min(x), max(x), 100)
poly_features = PolynomialFeatures(degree=selected_model_degree)
x_values_poly = poly_features.fit_transform(x_values.reshape(-1, 1))
y_values_pred = np.dot(x_values_poly, selected_model_coefficients)
plt.plot(x_values, y_values_pred, color='r', label=f'Degree={selected_model_degree}')
plt.title('Data and Selected Model')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
# END SOLUTION
```
